```python
drop_column = 8

try:
    n = grid[drop_column].index("*")
    n = n - 1
except:
    n = -1

grid[drop_column][n] = "*"

for n, column in enumerate(grid):
    print(n, column)
```

```python
def drop_block(column_number):
    try:
        n = grid[column_number].index("*")
        n = n - 1
    except:
        n = -1  
    grid[column_number][n] = "*"
```

```python
_ROWS = 7
_COLUMNS = 5

grid = [ [ "  " for i in range(_ROWS) ] for i in range (_COLUMNS) ]
```

```python
display_grid()
```

```python
_BLOCK = "\u2588\u2588"

def drop_block(column_number):
    try:
        n = grid[column_number].index(_BLOCK)
        n = n - 1
    except:
        n = -1  
    grid[column_number][n] = _BLOCK
```

```python
drop_block(1)
display_grid()
```

```python
def display_grid():
    the_columns = tuple(range(0, _COLUMNS))
    the_columns = tuple(range(0, _COLUMNS))
    for r in the_rows:
        print("|", sep="", end="")
        for c in the_columns:
            print(grid[c][r], "|", sep="", end="")
        print()    
```
